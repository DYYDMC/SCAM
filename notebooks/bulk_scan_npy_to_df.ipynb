```python
import numpy as np
```

```python
import pandas as pd
```

```python
data=[]
for ii in range(720):
    scan1 = np.load(f"./sky_scan{ii}.npy",allow_pickle=True)
    scan_dict =scan1.item()
    del scan1
    row={
        'wavelengths':scan_dict["wavelengths"],
        'spectrum':scan_dict["intensity"],
        'int_t':scan_dict["metadata"]['integration_time_ms'],
        'ts':scan_dict['ts']
    }
    data.append(row)
df = pd.DataFrame(data)
```

```python
df.to_pickle('./scan_20260120_150003.pkl')
```

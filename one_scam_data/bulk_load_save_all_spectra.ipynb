```python
import numpy as np
import pandas as pd
import os
import glob
```

```python
def build_df_from_npz(npz_path: str) -> pd.DataFrame:
    z = np.load(npz_path, allow_pickle=True)
    # keys = list(z.keys())
    # keys = [k for k in z.keys() if k != 'wavelength']
    exclude_cols = {'wavelength', 'motor_pan'}

    keys = [k for k in z.keys() if k not in exclude_cols]
    if not keys:
        raise ValueError("NPZ has no keys.")

    lengths = {}
    for k in keys:
        arr = z[k]
        lengths[k] = int(arr.shape[0]) if arr.ndim > 0 else 1
    n_rows = max(lengths.values())

    data: dict[str, Any] = {}
    for k in keys:
        arr = z[k]
        if arr.ndim == 0:
            data[k] = [arr.item()] * n_rows
        elif arr.shape[0] == n_rows and arr.ndim == 1:
            data[k] = arr
        elif arr.shape[0] == n_rows:
            data[k] = list(arr)
        else:
            data[k] = [arr] * n_rows
    return pd.DataFrame(data)
```
